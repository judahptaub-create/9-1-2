```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsRegressor

ct = ColumnTransformer(
    [("standardize", StandardScaler(), ["Gr Liv Area", "Bedroom AbvGr", "Full Bath", "Half Bath"],),
     ("dummify", OneHotEncoder(), ["Neighborhood"]),
    ],
    remainder="drop",
)

neighborhood_model = Pipeline(
    [("preprocessing", ct),
     ("lm", LinearRegression())]
)

X_train = df_ames[["Gr Liv Area", "Bedroom AbvGr", "Full Bath", "Half Bath", "Neighborhood"]]
y_train = df_ames["SalePrice"]

neighborhood_model.fit(X_train, y_train)
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(neighborhood_model,
                         X = X_train,
                         y = y_train,
                         scoring="neg_mean_squared_error",
                         cv=10)
scores
```

```python
import numpy as np
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import cross_val_score

X = df_ames[["Gr Liv Area", "Bedroom AbvGr", "Full Bath", "Half Bath", "Neighborhood"]]
y = df_ames["SalePrice"]

# Set up preprocessor (always standardize before KNN!)
ct = ColumnTransformer(
    [("standardize", StandardScaler(), ["Gr Liv Area", "Bedroom AbvGr", "Full Bath", "Half Bath"]),
     ("dummify", OneHotEncoder(handle_unknown="ignore"), ["Neighborhood"])
    ],
    remainder="drop"
)


k_values = [1, 3, 5, 7, 10, 15, 20, 25, 30, 50]
rmse_results = {}

for k in k_values:
    # Create pipeline with current k
    knn_pipeline = Pipeline([
        ("preprocessing", ct),
        ("knn", KNeighborsRegressor(n_neighbors=k))
    ])

    # Get negative MSE using 10-fold cross-validation
    scores = cross_val_score(
        knn_pipeline,
        X=X,
        y=y,
        scoring="neg_mean_squared_error",
        cv=10
    )

    rmse = np.sqrt(-scores.mean())
    rmse_results[k] = rmse

for k, rmse in rmse_results.items():
    print(f"k = {k:2d} -> Estimated Test RMSE: ${rmse:,.2f}")
```

```python
from sklearn.tree import DecisionTreeRegressor


X = df_ames[["Gr Liv Area", "Bedroom AbvGr", "Full Bath", "Half Bath", "Neighborhood"]]
y = df_ames["SalePrice"]


depths = [2, 3, 5, 7, 10, 15, 20, None]
tree_rmse_results = {}

for depth in depths:

    dt_pipeline = Pipeline([
        ("preprocessing", ct),
        ("tree", DecisionTreeRegressor(max_depth=depth, random_state=42))
    ])


    scores = cross_val_score(
        dt_pipeline,
        X=X,
        y=y,
        scoring="neg_mean_squared_error",
        cv=10
    )

    rmse = np.sqrt(-scores.mean())
    depth_label = "None (Unlimited)" if depth is None else str(depth)
    tree_rmse_results[depth_label] = rmse


for depth, rmse in tree_rmse_results.items():
    print(f"max_depth = {depth:>16s} -> Estimated Test RMSE: ${rmse:,.2f}")
```
